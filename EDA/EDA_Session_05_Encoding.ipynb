**pd.get_dummies**

- get dummies is a way to reprsent the categorical columns to numerical

- first it will check what are the labels in a column

- for example case status has two labels
    - Denied and Certified

- so it will create two new columns based on labels
    - col1: case_status_Denied
    - col2: case_status_Certified

**It is also called as One Hot Encoder**

- It makes new colums are independent each other
- Some of ML algorithms expect that columns should be **Independent** each other
