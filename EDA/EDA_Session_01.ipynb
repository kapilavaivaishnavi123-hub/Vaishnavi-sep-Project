**dtypes**

- every series is a key value pair
- every series is dictionary 
- In dictionary if we want keys: .keys()
- In dictionary if we want values: .values()

```python
visa_df.dtypes
```

```python
visa_df.dtypes.keys()
```

```python
visa_df.dtypes.values
```

```python
keys=visa_df.dtypes.keys().tolist()
values=visa_df.dtypes.values.tolist()
pd.DataFrame(zip(keys,values),columns=['Name','Data type'])
```

**very Important**
  
**- how to convert a series to df - dict keys tolist values then zip**

**- how to create a df whwn you have a list**

```python
df=pd.DataFrame(visa_df.dtypes).reset_index()
df.columns=['Names','Data types']
df
```

**task-2**

```python
visa_df.dtypes
# # idea
# # extract keys and values
# # apply the for loop on values
# # for i,j in zip(values,keys):
# #     if i=='object':
# #         j
keys=visa_df.dtypes.keys().tolist()
values=visa_df.dtypes.values.tolist()
cat,num=[],[]
for i,j in zip(keys,values):
    if j=='object':
        cat.append(i)
    else:
        num.append(j)
cat,num        
```
