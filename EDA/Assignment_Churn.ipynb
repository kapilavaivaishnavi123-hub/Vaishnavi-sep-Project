```python
sns.countplot(x="churn", data=df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
pd.crosstab(df["gender"], df["churn"])
```

```python
sns.countplot(x="gender", hue="churn", data=df)
plt.title("Gender vs Churn")
plt.show()
```

```python
bins = [0, 25, 35, 45, 55, 100]
labels = ["18-25", "26-35", "36-45", "46-55", "56+"]
df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)
```

```python
pd.crosstab(df["age_group"], df["churn"])
```

```python
sns.countplot(x="age_group", hue="churn", data=df)
plt.title("Age Group vs Churn")
plt.show()
```

```python
pd.crosstab(df["no_of_days_subscribed"], df["churn"])
```

```python
df.groupby("churn")["weekly_mins_watched"].mean()
```

```python
sns.boxplot(x="churn", y="weekly_mins_watched", data=df)
plt.title("Weekly Minutes Watched vs Churn")
plt.show()
```

```python
df.groupby("churn")["videos_watched"].mean()
```

```python
sns.boxplot(x="churn", y="videos_watched", data=df)
plt.title("Videos Watched vs Churn")
plt.show()
```

```python
sns.countplot(x="multi_screen", hue="churn", data=df)
plt.title("Multi-Screen Usage vs Churn")
plt.show()
```

```python
sns.boxplot(
    x="churn",
    y="maximum_days_inactive",
    data=df
)

plt.title("Inactive Days vs Churn")
plt.show()
```
